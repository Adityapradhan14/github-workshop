```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset('titanic')
print(titanic.head(10))
```

```python
sns.histplot(titanic['fare'], bins=30)
```

```python
sns.distplot(titanic['pclass'], bins=30)
```

```python
sns.jointplot(x= titanic['fare'], y = titanic['age'],kind='scatter')
```

```python
sns.rugplot(titanic['fare'])
```

```python
sns.barplot(titanic['fare'])
```

```python
sns.barplot(x='age', y='sex', data=titanic)
```

```python
sns.boxplot(x='age', y='sex', data=titanic)
```

```python
sns.barplot(x='age', y='sex', data=titanic,hue = "survived")
```

```python
sns.violinplot(x='age', y='sex', data=titanic)
```

```python
sns.stripplot(x='age', y='sex', data=titanic)
```

```python
sns.swarmplot(x='age', y='sex', data=titanic)
```

```python
# 1. Import required libraries
import pandas as pd
import numpy as np

# 2. Load the dataset
df = pd.read_csv('test.csv')

# 3. View first few rows
print(df.head())

# 4. Check for missing values
print("\nMissing Values:\n", df.isnull().sum())

# 5. Get initial statistics
print("\nInitial Statistics:\n", df.describe())

# 6. Check variable types
print("\nVariable Types:\n", df.dtypes)

# 7. Check dataframe dimensions
print("\nDataframe Dimensions:", df.shape)

# 8. Convert data types

df['Pclass'] = df['Pclass'].astype('category')

# 9. Normalize numeric columns
df['Age'] = (df['Age'] - df['Age'].min()) / (df['Age'].max() - df['Age'].min())
df['Fare'] = (df['Fare'] - df['Fare'].min()) / (df['Fare'].max() - df['Fare'].min())

# 10. Convert categorical columns to numeric
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
df['Embarked'] = df['Embarked'].map({'C': 0, 'Q': 1, 'S': 2})

# 11. Fill missing values in Embarked
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mean())

# 12. View final dataframe
print("\nFinal DataFrame after preprocessing:\n", df.head())
```
